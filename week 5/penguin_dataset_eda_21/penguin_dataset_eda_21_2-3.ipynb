Biological Measurement KPIs

4.	Average Body Mass (g)

-	Overall average
-	Average by species

-	Average by gender


```python
df.columns
```

```python
overall_average_bodymass=df['body_mass_g'].mean()
bodymass_average_species=df.groupby("species")['body_mass_g'].mean()
bodymass_average_gender=df.groupby("sex")['body_mass_g'].mean()
```

```python
print("overall average bodymass:")
print(overall_average_bodymass)
print("Average body mass by species:")
print(bodymass_average_species)
print("Average body mass by Gender:")
print(bodymass_average_gender)
```

5.	Average Flipper Length (mm)

-	By species
-	By gender

```python
print(overall_average_flipper)
print(flipper_average_species)
print(flipper_average_gender)
```
