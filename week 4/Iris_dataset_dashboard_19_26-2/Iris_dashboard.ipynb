```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
iris = load_iris()

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target
df['species'] = df['species'].map({0:'Setosa',1:'Versicolor',2:'Virginica'})

df.head()
```

KPI 1 — TOTAL RECORDS

```python
total_records = df.shape[0] #[0]--> rows
print("Total Records:", total_records)
```

KPI 2 — SPECIES DISTRIBUTION

```python
species_count = df['species'].value_counts()
species_percent = df['species'].value_counts(normalize=True)*100

print("\nSpecies Count:\n", species_count)
print("\nSpecies Percentage:\n", species_percent)
```

KPI 3 — AVERAGE FEATURE VALUES BY SPECIES

```python
avg_features = df.groupby('species').mean()
print("\nAverage Feature Values by Species:\n", avg_features)
```

KPI 4 — CORRELATION MATRIX

```python
corr_matrix = df.drop('species',axis=1).corr()
print("\nCorrelation Matrix:\n", corr_matrix)
```

```python
correlation_matrix=df.corr(numeric_only=True)
correlation_matrix
```

KPI 5 — OUTLIER DETECTION (IQR METHOD)

```python
def detect_outliers(data):
    outlier_summary = {}

    for col in data.columns:
        Q1 = data[col].quantile(0.25)
        Q3 = data[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5*IQR
        upper = Q3 + 1.5*IQR

        outliers = data[(data[col] < lower) | (data[col] > upper)]
        outlier_summary[col] = len(outliers)

    return outlier_summary

outliers = detect_outliers(df.drop('species',axis=1))
print("\nOutlier Summary:\n", outliers)
```

Static dashboard using plotly:

```python
# IMPORT LIBRARIES
import pandas as pd
import plotly.express as px
from sklearn.datasets import load_iris

# LOAD DATASET
iris = load_iris()

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target
df['species'] = df['species'].map({0:'Setosa',1:'Versicolor',2:'Virginica'})

# KPI CALCULATIONS
total_records = len(df)

species_count = df['species'].value_counts().reset_index()
species_count.columns = ['species','count']

avg_features = df.groupby('species').mean().reset_index()

corr_matrix = df.drop('species',axis=1).corr()

# PIE CHART — SPECIES DISTRIBUTION
fig1 = px.pie(
    species_count,
    names='species',
    values='count',
    title='Species Distribution'
)

# BOXPLOTS — FEATURE COMPARISON
fig2 = px.box(df, x='species', y='sepal length (cm)', title='Sepal Length by Species')
fig3 = px.box(df, x='species', y='sepal width (cm)', title='Sepal Width by Species')
fig4 = px.box(df, x='species', y='petal length (cm)', title='Petal Length by Species')
fig5 = px.box(df, x='species', y='petal width (cm)', title='Petal Width by Species')

# PAIR PLOT — SCATTER MATRIX
fig6 = px.scatter_matrix(
    df,
    dimensions=df.columns[:-1],
    color='species',
    title='Pair Plot'
)

# CORRELATION HEATMAP
fig7 = px.imshow(
    corr_matrix,
    text_auto=True,
    title="Correlation Heatmap"
)

# SHOW ALL
fig1.show()
fig2.show()
fig3.show()
fig4.show()
fig5.show()
fig6.show()
fig7.show()

# PRINT KPIs
print("Total Records:", total_records)
print("\nSpecies Distribution:\n", species_count)
print("\nAverage Feature Values by Species:\n", avg_features)
print("\nCorrelation Matrix:\n", corr_matrix)
```
