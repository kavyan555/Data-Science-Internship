Chart requiremnets

```python
plt.figure()
avg_mpg_year = df.groupby("model year")["mpg"].mean()
sns.lineplot(x=avg_mpg_year.index, y=avg_mpg_year.values)

plt.title("Fuel Efficiency Trend")
plt.xlabel("Model Year")
plt.ylabel("Average MPG")
plt.show()
```

```python
plt.figure()
sns.barplot(data=df, x="origin", y="mpg")

plt.title("Average MPG by Origin")
plt.xlabel("Origin")
plt.ylabel("Average MPG")
plt.show()
```

```python
plt.figure()
df["origin"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Market Share by Origin")
plt.ylabel("")
plt.show()
```

```python
plt.figure()
sns.regplot(data=df, x="weight", y="mpg")

plt.title("MPG vs Weight")
plt.xlabel("Weight")
plt.ylabel("MPG")
plt.show()
```

```python
plt.figure()
sns.regplot(data=df, x="horsepower", y="mpg")

plt.title("MPG vs Horsepower")
plt.xlabel("Horsepower")
plt.ylabel("MPG")
plt.show()
```

```python
plt.figure()
sns.countplot(data=df, x="cylinders")

plt.title("Engine Distribution")
plt.xlabel("Cylinders")
plt.ylabel("Vehicle Count")
plt.show()
```

```python
plt.figure()
sns.histplot(df["mpg"], bins=15, kde=True)

plt.title("MPG Distribution")
plt.xlabel("MPG")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure()

corr = df[["mpg","weight","horsepower",
           "displacement","cylinders","acceleration"]].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Matrix")
plt.show()
```
