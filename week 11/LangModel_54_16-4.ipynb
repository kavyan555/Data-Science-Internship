```python
Types of language models:
1.statistical models:

n-gram model:

it uses previous words to predict next word

examples:
bigram---uses previous words token
unigram---uses previous word token
```

```python
2.neural language models:

RNN(recurrent neural networks):

remebers the previous words

mainly it is used for the sequential data

it faces with vanishing gradient problem

LSTM(long short term memory):

it is used to handle long term dependencies

advance version of rnn
```

```python
3.Transformer models:

it is the advance version of nlp
where it has the self attention mechasim
mainly it focusses on important words in a sentence
```

```python
4.large language models:
llms are advance language models which is used to train on huge amount of the data

examples:
gpt(generative pre trained transformers)
bert(bidirectional encoding representative transformer)
gemini
```

```python
5.how large language models work:

1.embeddings:
it helps to convert the textual data into vectors(numbers)

2.positional encoding:
it helps to understand word order

3.self attention:
it helps to find the realtionship between the words

4.feed forward layers:
it helps to learn complex patterns inside the data

5.decoding:

it helps to generate the outputs word by word
```
