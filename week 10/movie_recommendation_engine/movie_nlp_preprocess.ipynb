```python
import ast

def extract_names(obj_str,key="name",max_items=None):
    try:
        items=ast.literal_eval(obj_str)
        if not isinstance(items,list):
            return []
        return[
        item[key].replace(" ","").lower()
        for item in items[:max_items]
        if isinstance(item,dict)
        and key in item
        and isinstance(item[key],str)
        ]
    except(ValueError,SyntaxError):
        return []
```

```python
df['genres']=df['genres'].apply(extract_names)
```

```python
import ast

def extract_director(crew_str):
    try:
        crew = ast.literal_eval(crew_str)
        for member in crew:
            if member.get('job') == 'Director':
                return [member['name'].replace(" ", "").lower()]
    except:
        return []
    return []
```

```python
df['crew']=df['crew'].apply(extract_director)
```

```python
df['overview']=df['overview'].astype(str).str.lower().str.split()
```

```python
df['tags']=df['overview']+df['genres']+df['keywords']+df['cast']+df['crew']
```

```python
df['tags']=df['tags'].apply(lambda x:" ".join(x))
```

```python
final_df=df[['id','title','tags']]
```

```python
final_df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cv=CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors=cv.fit_transform(final_df['tags']).toarray()
```

```python
similarity=cosine_similarity(vectors)
```

```python
def recommend(movie_title):
    movie_title=movie_title.lower()
    matches=final_df[final_df['title'].str.lower()==movie_title]
    if matches.empty:
        return "movie not found"

    movie_index=matches.index[0]
    distances=similarity[movie_index]
    movie_list=sorted(
    list(enumerate(distances)),
    reverse=True,
    key=lambda x:x[1])[1:10]
    return[final_df.iloc[i[0]].title for i in movie_list]
```

```python
recommend('Spectre')
```
