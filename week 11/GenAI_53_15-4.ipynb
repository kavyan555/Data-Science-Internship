```python
Generative ai core components:

1.data:

to build any models in gen ai like chat gpt we require to train our model with huge amount of the
data so that it can learn the patterns from it to generate the new content

examples:

text:datasets(like bools,articles,blogs,wikipedia etc)
images data
audios data
videos data
```

```python
2.model architecture:
this helps to define how ai model learns the patterns from the data

popular models:

transformers
gans
vaes
gpt
bert
```

```python
3.training:
it is the process where the model learns the patterns from the data

steps:

collect the datasets
preprocess the data
train the model
optimization

techniques:
supervised learning,rlfh(reinforcement learning from human feedback)
```

```python
4.prompt:

gen ai models will work based on the instructional prompts

prompt--input given to the llm to get the response
```

```python
5.inference engine:

it is the process of generating the output by using the trained model

when user send the prompt to the llm:

the model will process the prompt
predict the next tokens
generate the response
```

```python
6.output generation:

the model generates the new content based on the learned patterns

text
images
code
video
audio
```

```python
7.fine tuning:
it adaps the pretrained model based on the specific domain

sft
lora
qlora
peft
```

```python
in simple terms what we say about gen ai is:
it is used to load the data and choose any one model for the training process and it takes
some instructions how the output should be(Structure) and reaching to the output and finetuning
the model based on domain specific knowledge and reach to the deployement to look for the ui 
interface where the user pass the prompt reach to llm model and generate the response
```

```python
gen ai architecture:

for building any of the applications of gen ai like:
chatbots/summarization/recommendation engines/rag application
```

```python
1.user interface---chatbot/webapp
2.Application layer
3.prompt engineering
4.embedding model
5.vector database
6.rag(retrieval augumented generation)
7.llm(large language model)
8.post processing
9.response to the user
```

```python
1.user interface/user interaction:
it starts when the user send a request or prompt to the llm model
where this layer helps to collect the user prompt and send to the backend to reach the output

examples:
explain about nlp?

accepts the prompt
send the request to the llm
displays the llm response
```

```python
2.Application layer:

acts as bridge between the frontend and the llm model
```

```python
3.prompt engineering:

the user query is converted to the structure format for the llm model to generate the response
```

```python
4.embedding model:
it is mainly used to convert text information into numerical representations(vectors) where
the machines can understand to perform different tasks
```

```python
5.vector data base:

it is mainly used to store the data into vector db to retrive the similar information

1.converts the query(embedding)
2.search the similar vectors from the vector db
3.it retrive the relevant information
```

```python
6.rag(retrieval augumented generation):

it is mainly used to connect with external data sources

where it helps to retrieve the relevant knowledge from the databases
```

```python
7.large language models:

it is one of the core intelligence of the system

where it will helps:

to read the user query
to retrieve the documents
to give the prompt instructions

ex:

gpt
gemini
grok
perplexity
claude
llama
```

```python
8.post processing:

after the llm model is generating the output
it will helps how the structure of the output and reach to the response to the user interface
```

```python
9.response to the user:
generated response for the user query from the llm model
```
