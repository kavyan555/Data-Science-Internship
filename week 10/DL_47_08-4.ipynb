```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt
```

```python
(x_train,y_train),(x_test,y_test)=keras.datasets.mnist.load_data()
```

```python
model=keras.Sequential([
    keras.layers.Dense(128,activation='relu',input_shape=(784,)),
    keras.layers.Dense(64,activation='relu'),
    keras.layers.Dense(10,activation='softmax',)
])
```

```python
model.compile(optimizer='adam',
              loss='SparseCategoricalCrossentropy',
              metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=5)
```

```python
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f'\nTest accuracy: {test_acc}')
```

```python
predictions=model.predict(x_test)
predicted_labels=np.argmax(predictions,axis=1)
```

```python
import tensorflow as tf
from tensorflow.keras import datasets,layers,models
import matplotlib.pyplot as plt
import numpy as np
```

```python
(x_train,y_train),(x_test,y_test)=datasets.mnist.load_data()
```

```python
x_train=x_train/255.0
x_test=x_test/255.0
```

```python
x_train
```

```python
x_train.ndim
```

```python
x_train=x_train.reshape(-1,28,28,1)
x_test=x_test.reshape(-1,28,28,1)
```

```python
x_train.ndim
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
model=models.Sequential([
    layers.Conv2D(32,(3,3),activation='relu',input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(64,(3,3),activation='relu'),
    layers.MaxPooling2D((2,2)),

    layers.Flatten(),
    layers.Dense(64,activation='relu'),
    layers.Dense(10,activation='softmax')
])
```

```python
model.compile(optimizer='adam',
              loss='SparseCategoricalCrossentropy',
              metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=5)
```

```python
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f'\nTest accuracy: {test_acc}')
```

```python
predictions=model.predict(x_test)
predicted_labels=np.argmax(predictions,axis=1)
```

```python
plt.figure(figsize=(12,6))

for i in range(10):
    plt.subplot(2,5,i+1)
    img=x_test[i].reshape(28,28)
    plt.imshow(img,cmap='gray')
    plt.title(f"Actual:{y_test[i]}\n predicted:{predicted_labels[i]}")
    plt.axis('off')
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,SimpleRNN,Embedding
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
text=[
    "i love data science",
    "i love machine learning",
    "deep learning is powerful",
    "rnn is useful for sequential data",
    "lstm solve vanishing gradient problem"
]
```

```python
tokenizer=Tokenizer()
tokenizer.fit_on_texts(text)
```

```python
word_index=tokenizer.word_index
total_words=len(word_index)+1
print(word_index)
```

```python
input_sequences=[]
for line in text:
    token_list=tokenizer.texts_to_sequences([line])[0]
    for i in range(1,len(token_list)):
        n_gram=token_list[:i+1]
        input_sequences.append(n_gram)
```

```python
print(input_sequences)
```

```python
max_len=max(len(seq) for seq in input_sequences)
```

```python
input_sequences=pad_sequences(input_sequences,maxlen=max_len,padding='pre')
```

```python
x=input_sequences[:,:-1]
y=input_sequences[:,-1]
```

```python
x
```

```python
y
```

```python
model_rnn=Sequential([
    Embedding(total_words,10,input_length=max_len-1),
    SimpleRNN(50),
    Dense(total_words,activation='softmax')
])
```

```python
model_rnn.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model_rnn.fit(x,y,epochs=100)
```

```python
def predict_next_word(model,tokenizer,text,max_len):
    token_list=tokenizer.texts_to_sequences([text])[0]
    token_list=pad_sequences([token_list],maxlen=max_len-1,padding='pre')

    predicted=np.argmax(model_rnn.predict(token_list),axis=1)

    for word,index in tokenizer.word_index.items():
        if index==predicted:
           return word

```

```python
predict_next_word(model_rnn,tokenizer,"rnn is",max_len)
```

```python
predict_next_word(model_rnn,tokenizer,"lstm solve vanishing",max_len)
```
