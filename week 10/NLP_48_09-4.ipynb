```python
natural language processing:
it is a branch of ai which is used to understand and intepret and generate the tasks in a human
language

it is a branch of ai which helps to understand the human language and also helps to understand
the meaning and to generate human-like text

with the help of nlp we can perform different text related tasks
```

```python
it helps to combine:
linguistics
computer science
Artificial intelligence

ex:
"i love this product"

machine understands like

emotion----positive
intent---product feedback
output--->sebtiment analysis---->positive
```

```python
Types of nlp tasks:

1.natural language understanding(NLU):

machine understands the text to perform the use cases like:

sentiment analysis
intent detection
entity recognition

2.natural language generation(NLG):

machine understands the text to perfrom the use cases like:

chatbots
story generation
content creation
text summarization

spam detection----->NLU(natural language understanding)

chatgpt response---->NLG(natural language generation)
```

```python
use cases:

1.chatbots
>>customer support chatbots
>>banking assistants

examples:
google assistant
alexa

2.sentiment analysis
>>product review
>>social media analysis

examples:
flipkart/amazon/instagram etc

3.machine translation:

it helps to convert one language to another language

examples:

google translate

4.search engines:

understanding the use queries

5.spam detection:

email classification

examples:

spam filtering
```

```python
challanges in nlp:

1.noisy data:

typos
emojis
slang

2.sarcasm detection:

its very hard for the machines to understand

3.context understanding:

same sentence--->different meaning

example:

"he is cold"

emotion
temperature
```

```python
4.Ambiguity:

word has multiple meanings

example:

"bank"

financial bank
river bank
```

```python
life cycle of nlp: 
1.text input---->raw data examples: reviews tweets emails 
2.text preprocessing: clean the data remove noise tokenize the data
3.feature extraction: converting the textual information--->numerical(vectors) tf-idf: word embeddings 
4.model building: train ml/dl model examples: Logistic regression bert(bidirectional encoding representative transformer) 
5.evaluation: checking the performance of the model Accuracy/f1-score 
6.deployement: Api/web app/gradio/flask/fast api/streamlit
```

```python
text preprocessing:

it is the process where we can clean,transform the raw data into structured format
```

```python
Techniques of nlp:

1.tokenization:

it helps to split the textual data into smaller pieces

ex:["what is machine learning life cycle"]

['what','is','machine','learning','life','cycle']
```

```python
2.text preprocessing:

we can convert the data into lower case

we can remove the stop words(is,the,a etc)

we can also remove the punctuations for better analysis

stemming and lemmatization

we can also reduce the word into their root word
```

```python
3.parts of speech(pos)

we can easily identify the grammatical categories like noun,pronoun,adverb etc
```

```python
4.NER(named entity recognition)
peoples
places
organizations
objects
companies
locations
```

```python
5.word embeddings:
word2vec
glove
bert
roberta
t5
```

```python
6.tf-idf(term frequency-inverse document frequncy)

improve bag of words  by reducing the common words

search engines
text classification
```

```python
7.bag of words(bow):
it helps to show the count of word frequency
```

```python
NNP → Proper noun, singular
Example: “Nlp”, “John”, “London”
VBZ → Verb, 3rd person singular present
Example: “makes”, “runs”, “eats”
NNS → Noun, plural
Example: “machines”, “dogs”, “cars”
VB → Verb, base form
Example: “understand”, “eat”, “run”
DT → Determiner
Example: “the”, “a”, “an”
JJ → Adjective
Example: “human”, “big”, “fast”
NN → Noun, singular or mass
Example: “language”, “apple”, “water”
```

```python
from gensim.models import Word2Vec
```

```python
sentences=[
    ["i","love","nlp"],
    ["nlp","is","fun"],
    ["i","enjoy","learning"]
]
```

```python
model=Word2Vec(sentences,vector_size=50,window=3,min_count=1)
```

```python
model.wv['enjoy']
```

```python
model.wv.similarity('nlp','fun')
```

```python
model.wv.similarity('nlp','learning')
```
