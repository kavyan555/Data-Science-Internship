```python
transformer:
it is a advance version of natural language processing
it is also called as neural network model which is used to work on the sequential data

problems seen in rnn/lstm:
long dependencies
it is very low to process the sequential data
it is not scalable

transformers will be helpful to solve:
parallel processing
it is highly scalable because of its models like bert,gpt etc
capture the long range dependencies
```

```python
transformer architecture:

input data---->encoder---->context----->decoded---->output
```

```python
core components of transformers:

self attention mechanism
positional encoding
multi head attenttion
encoder-decoder
embeddings
tokenization
positional wise feed forward neural network
```

```python
self attention mechanism:

QKV----query,key,value

Attention(Q,K,V)=softmax(QK**t/square root dk)V
```

```python
positional encoding:
it adds the information about the position of each and every token in a sequence
```

```python
Multi-head attention :
it  extends self-attention by splitting the input into multiple heads,
enabling the model to capture diverse relationships and patterns.

encoder
decoder
```

```python
encoder-decoder:
This layer lets the decoder attend over the encoders output.
It helps the decoder to align and focus on the appropriate input tokens
when generating each output token, enabling sequence-to-sequence tasks like translation.
```

```python
embeddings:

it is used to convert textual data into numerical representations(vector) is called as embeddings


Embeddings are widely used because they represent data in a meaningful

and efficient way, helping models understand relationships and perform better across tasks.

Capture semantic relationships by placing similar items closer in vector space.

Reduce dimensionality while preserving important patterns and features.

Support transfer learning by reusing embeddings across different tasks.

Automatically extract useful features, reducing manual effort.
    
Provide interpretable representations through distances and directions between vectors
```

```python
tokenization:

it is the process of converting the sentences into small and managable pieces like words,sub words
and characters such types are called as tokens

"i love learning transformers"


["i","love","learning","transformers"]
```

```python
from transformers import AutoTokenizer
tokenizer=AutoTokenizer.from_pretrained("bert-base-uncased")
text="i love learning transformers"

tokens=tokenizer.tokenize(text)
```

```python
print(tokens)
```

```python
positional wise feed forward neural network:

This forward pass assumes the data has been passed through the embedding,
positional encoding, and multi-head attention layers. It does not use layer 
normalization or residual addition, which will be implemented before and after 
this network in the Encoder
```

```python
from transformers import pipeline

qa = pipeline("question-answering")

context = "Transformers are models that process sequential data using attention mechanisms."

question = "What do transformers use?"

result = qa(question=question, context=context)
print(result)
```

```python
from transformers import pipeline

ner = pipeline("ner", grouped_entities=True)

text = "Elon Musk founded SpaceX in the United States."
print(ner(text))
```

```python
from transformers import pipeline

summarizer = pipeline("summarization")

text = """Transformers have revolutionized natural language processing by enabling models
to process words in relation to all other words in a sentence, rather than sequentially."""

print(summarizer(text, max_length=30, min_length=10))
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification 
import torch

tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased") 
model = AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased")

inputs = tokenizer("I love Transformers!", return_tensors="pt") 
outputs = model(**inputs)

logits = outputs.logits 
predicted_class_id = torch.argmax(logits).item()

print(predicted_class_id)
```
