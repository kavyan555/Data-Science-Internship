```python
prompt engineering:

it is used to give proper input or proper instructions to the large language models like gpt
to generate the accurate and relevant response
```

```python
features of prompt engineering:
standarization:it is helpful for building some prompt templates while working with teams/projects
safety:no confidential data will leak
integration:combine the prompt with apis and also to work with automation pipelines
optimization:which helps to improve the accuracy
```

```python
1.clear intent specification:
what you want,in which format you want,which method you want to work with

ex: explain ai

ex:explain ai in simple terms with three real time examples

2.role assignment:

we need to tell the model how it should act as

ex:you are senior data analyst explain about exploratory data analysis in detail explanation

3.iterative prompting(refinement loop):

start prompt---->refine---->expected output

4.structure of the output:

it helps to give the output in the structure what ever we are expecting like pdf,json,txt,etc

5.context enrichment:

based on the contextual information which we have provided

6.automation & rooling : langchain and llama index
```

```python
how to craft effective prompting:

RICE(role,instruction,context,explicit o/p)

Role:what should the model acts as ?

example: act as a python developer, act as a data analyst, act as a sql developer

instruction:what should it do?

example:write a code for displaying patterns by using loopings

context:it is used to add extra details about the code

example:use this code and give me the easiest way to understand the patterns how it is working

explicit output: which defines the structure of the output

example:give me the entire code in .py in a downloadable format
```

```python
"you are a sql expert generate a query of kpi requirements which was provided in the dataset provide
only sql code without any explanation"
```

```python
types of prompting:
1.zero shot prompting
2.one shot prompting
3.few shot prompting
4.instruction based prompting
5.role based prompting
6.chain of thought prompting
7.self consistency prompting
8.reAct prompt(reasoning+act)
```

```python
1.zero shot prompting:

providing only instructions without any examples

use cases:

quick answers
general knowledge

example:

explain about machine learning in simple terms
```

```python
2.one-shot prompting:
giving one example model to guide the model how exactly it should give the results

example:

input:Apple
output:Fruit

input:potato
output:?

this type of prompting will helps the model to understand the patterns very well
```

```python
3.few shot prompting:

giving more than one example to recognize the pattern very well

examples:

input: apple---->fruit
       carrot---->vegetable
       banana----->fruit
       radish----->?
```

```python
4.instruction based prompting:

giving the llm model clear instructions to generate accurate results

example:

write a professional email to request leave for 2 days
```

```python
5.role based prompting:

Assigning the role to the llm model how should it act as

example:

you are a data scientist explain about loss function in machine learning
```

```python
6.chain-of-thought:

Ask the model to think step by step

example:

if a train travels from vizag to guntur how many hours it takes to reach if it travels 60 km /hou
```

```python
7.self consistency prompting:

it helps the model to generate multiple reasoning paths to pick the best

example:explain about ai in two responses i will pick any one from them
```

```python
8.reAct prompting:

reason+act

which helps to combine reasoning+actions

search---->analyze----answer

mainly used in 

ai agents

chatbots with tools
```
