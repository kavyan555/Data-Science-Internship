```python
#using the list slicing

def chunk_list(data,chunk_size):
    for i in range(0,len(data),chunk_size):
        yield data[i:i+chunk_size]

data=list(range(1,10))
for chunk in chunk_list(data,3):
    print(chunk)
```

```python
#using list comprehension:

data=list(range(1,10))
chunks=[data[i:i+3] for i in range(0,len(data),3)]
print(chunks)
```

```python
def chunk_iterable(iterable,size):
    it=iter(iterable)
    while True:
        chunk=list(islice(it,size))
        if not chunk:
            break
        yield chunk
```

```python
5.recursive chunking:
if the data is too big to process then we will split the data again
```

```python
def recursive_chunk(text,size):
    if len(text)<=size:
        return[text]
    return[text[:size]]+recursive_chunk(text[size:],size)
```

```python
Hallucinations (in AI and especially Large Language Models) refer to outputs that sound confident 
but are incorrect, fabricated, or not grounded in real data.
```

```python
1. Factual Hallucination

Definition: The model generates incorrect factual information.

Example:

 “The capital of Australia is Sydney.”
 Correct: Canberra
```

```python
2.Semantic Hallucinations

Definition:
Occurs when the model misunderstands the meaning (semantics) of the input
and produces a response that is linguistically correct but conceptually wrong.

Example:

You: “What is Python?”
AI: “Python is a big snake.”
 Mistake: It didn’t understand you meant programming language
```

```python
3. Reasoning Hallucination

AI understands the topic, but thinks wrong

Example:

Statement: “All apples are fruits. Banana is a fruit.”
AI says: “So banana is an apple.” 

Mistake: Wrong logic
Just because both are fruits doesn’t mean they are the same
```
