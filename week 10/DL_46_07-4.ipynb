```python
Types of deep learning:

FNN(Ann)
CNN
RNN
LSTM
GRU
GAN
VAE
Transformers
```

```python
1.Ann(Aritificial neural network/feed forward neural network)

structured data(excel/tables)

input data--->hidden layer---->output

it is mainly used to perform simple tasks like classification and regression models

examples:

spam detection
house price prediction
```

```python
2.CNN(convolutional neural networks):

it helps to work with image related data

Architecture of cnn:

1.convolution layers

it helps to detect the shapes,texture,color and edges

2.Relu(rectified linear unit):

it acts as a activation functions for non linear transformations

3.pooling layers:

it helps to reduce the sizes

4.flatten layer:

it is used to convert the images into vectors

5.final decision:

we get the output data(prediction)
```

```python
3.RNN(recurrent neural networks):

it is used to solve the sequential data

it have one disadvantage i.e vanishing gradient problem

it will not able to remember the previous info
```

```python
4.LSTM(long short term memory):

it is the advance version of rnn

to avoid the vanishing gradient problem we use lstm

lstm consists of different gates:

1.forget gate---it helps to remove useless information

2.input gate----helps to add new information

3.output gate----it gives the output
```

```python
5.GRU(gated recurrent unit):

it is also act as lstm

it will perform very good and gives faster results
```

```python
6.GAN(generative adversarial neural network):
it consists of two neural networks:

generator----creates fake data

discriminator----checks real/fake

work flow:

generator will create the fake image

discrimator will checks whether the image is real/fake

feedback loop: where it helps to improve the generator image
```

```python
7.VAES(variational auto encoders):

it is the improvement of auto encoder

autoencoder:consists of two neural networks i.e encoder and decoder

encoder: will compress the data
decoder:reconstructs the data

VAES:
it learns from the data distributions:
how the data was spread between mean and std
it is mainly helpful to generate the new realistic data
```

```python
8.Transformers:

the modern form of natural language processing

it is mainly used to handle the sequential data

it follows the self attention rule
```
